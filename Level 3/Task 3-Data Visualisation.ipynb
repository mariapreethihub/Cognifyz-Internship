```python
for col in df.select_dtypes('object'):
    df[col]=df[col].str.replace(r'[^a-zA-Z0-9,]+',"",regex=True)        # Removing special characters from the dataset
```

```python
df['Cuisines']=df['Cuisines'].fillna(df['Cuisines'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Cuisines']=df['Cuisines'].str.split(",")             # Cuisines are stored as single string in field 'Cuisines' it is split using split()
df=df.explode('Cuisines')                               # For each cuisine separate row is created using explode()
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns=['Restaurant ID','Switch to order menu','Rating color','Rating text','Currency','Locality Verbose'],inplace=True)
```

```python
df['Aggregate rating'].value_counts()
```

```python
df=df[df['Aggregate rating']!=0]     # Aggregate rating with value 0 is removed as objective of task is to predict aggregate rating and target value with 0 is not useful.
```

```python
df.columns
```

# Distribution of rating

```python
sns.boxplot(x=df['Aggregate rating'], color="green")
plt.title("Boxplot of Aggregate Ratings")

plt.show()
```

# Comparing rating with different features

### Average rating Vs Table booking,online delivery and delivering now

```python
features=['Has Table booking','Has Online delivery','Is delivering now']
plt.figure(figsize=(8,6))
for index,col in enumerate (df[features].columns,1):
    plt.subplot(2,2,index)
    sns.barplot(data=df,x=col,y='Aggregate rating',color='red',weights=5)
    plt.title(f"{col} Vs Aggregate Rating")

plt.tight_layout()
plt.show()
```

Comment :

- Restaurants that offer table booking have higher average ratings compared to those that don’t.
- The possible reason could be customers convenience, reduces waiting time and improves dining experience.
- Restaurants with online delivery have slightly lower ratings than dine-in restaurants.This could be due to delivery time and temperature of food which may affect customer satisfaction.
- Service features influence customer ratings.Table booking has strong value, while delivery needs improvement.


### Restaurant and average rating

```python
avg_rating =df.groupby('Restaurant Name')['Aggregate rating'].mean().sort_values(ascending=False).reset_index()
avg_rating.columns=['Restaurant Name','Average Rating']
avg_rating.head(50)
```

### Average rating and cuisines

```python
top5_cuisines=df['Cuisines'].value_counts().head(5).index
top5_cuisines
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df[df['Cuisines'].isin(top5_cuisines)],x='Cuisines',y='Aggregate rating',ci=None,estimator='mean',color='orange')
plt.title("Top 5 Cuisines-Avg rating")
plt.ylabel("Average rating")
plt.show()
```

Comment :
- Italian cuisine has highest average rating.

### City Vs Rating

```python
df.columns
```

```python
df['City'].unique().tolist()
```

```python
top5=df.groupby('City')['Aggregate rating'].mean().reset_index().sort_values(by='Aggregate rating',ascending=False).head()
top5
```

```python
plt.pie(top5['Aggregate rating'], labels=top5['City'],autopct='%1.1f%%',colors = ['pink', 'skyblue', 'lightgreen', 'green', 'violet'])
plt.title("Top 5 Cities by Total Ratings")
plt.show()
```

Comment :
- Inner city has the highest average rating.
- Top 5 cities have higher average rating than other cities which suggests that customer in these place may expereince better dine in experience.
